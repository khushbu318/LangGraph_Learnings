```python
from langgraph.store.memory import InMemoryStore
```

```python
# creating a object of memory store
store = InMemoryStore()
```

```python
## creating a namespace 
## it is similar as the google drive  to organize the folder and files
## namespace (users, U1) or (users.u2) or (users.u1.profile)
# so folder users have folder u1 and u2
# namespace is a tuple of strings, where each string is a folder name and using , we cna crate 

namespace = ('user', 'u1')
```

```python
## put method is user to store the date in the namespace, it takes 3 parameters namespace, key and value

# adding memories to the namespace
store.put(namespace, '1', {'data':'user linkes pizza'})
store.put(namespace, '2', {'data': 'user prefer dark mode'})
```

```python
# anothere namespace

namespace2 = ('user', 'u2')

# add memories to the namespace2
store.put(namespace2, '1', {'data':'User likes burger'})
store.put(namespace2, '2', {'data': 'User prefer light mode'})
```

```python
## get method is used to retrieve the data from the namespace, it takes 2 parameters namespace and key

## need key to use the key
store.get(namespace, '1')  # it will return the data stored in the namespace with key '1'
```

## Retrieving all Memories

```python
# search method is user to retrieve all the data from the namespace

items = store.search(namespace)

for item in items:
    print(item.value)
```

### Semantic search
- Search on the basis of meaning

```python
from langchain_openai import ChatOpenAI
import os

from langchain_huggingface import HuggingFaceEmbeddings

embeddings_model = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True},
)
```

```python
store = InMemoryStore(index={'embed': embeddings_model, 'dims':1536})
```

```python
## createing namespace for the user to do the semantic search
namespace1 = ('user', 'u1')
```

```python
store.put(namespace1, '1', {'data': 'User prefers light mode'})
store.put(namespace1, '2', {'data': 'User prefers concise responses'})
store.put(namespace1, '3', {'data': 'User likes email notifications'})
store.put(namespace1, '4', {'data': 'User uses metric units'})
store.put(namespace1, '5', {'data': 'User prefers English language'})
store.put(namespace1, '6', {'data': 'User likes weekly summary emails'})
store.put(namespace1, '7', {'data': 'User prefers 24-hour time format'})
store.put(namespace1, '8', {'data': 'User prefers markdown formatted responses'})
store.put(namespace1, '9', {'data': 'User likes keyboard shortcuts'})
store.put(namespace1, '10', {'data': 'User prefers automatic cloud backups'})
```

```python
items = store.search(namespace1, query='What time format does the user prefer?', limit=3)

# What theme does the user prefer?
# How does the user like responses to be formatted?
# How verbose should my replies be?
# What notification preferences does the user have?
# Which measurement system should I use when answering?
# What language should I respond in?
# What time format does the user prefer?
# Does the user prefer keyboard shortcuts?
# Should I enable automatic backups for this user?
# What communication preferences has the user saved?
# What UI preferences has the user configured?
# What formatting preferences should I follow when generating responses?
# Tell me everything I know about this user's preferences.
# What user preferences are stored in namespace1?


for item in items:
    print(item.value)
```
